```python
def obtener_info_album(album_id):
    url_album = f"https://api.deezer.com/album/{album_id}"
    response_album = requests.get(url_album)

    if response_album.status_code != 200:
        return {
            "album_id": album_id,
            "release_year": None,
            "genre_id": None,
            "genre_name": None
        }

    data_album = response_album.json()

    release_date = data_album.get("release_date")

    if release_date:
        release_year = release_date[:4]
    else:
        release_year = None

    genres = data_album.get("genres", {}).get("data", [])

    if genres:
        genre_id = genres[0].get("id")
        genre_name = genres[0].get("name")
    else:
        genre_id = None
        genre_name = None

    return {
        "album_id": album_id,
        "release_year": release_year,
        "genre_id": genre_id,
        "genre_name": genre_name
    }
```

```python
albumes_info = []

for album_id in album_ids_unicos:

    info_album = obtener_info_album(album_id)

    albumes_info.append(info_album)

    time.sleep(0.2)
```

```python
def obtener_tracks_artista(nombre_artista):
    # 1. Buscar artista
    url_artist = "https://api.deezer.com/search/artist"
    params_artist = {"q": nombre_artista}

    response_artist = requests.get(url_artist, params=params_artist)
    data_artist = response_artist.json()

    if not data_artist["data"]:
        print(f"No se encontró artista: {nombre_artista}")
        return pd.DataFrame()

    artist_id = data_artist["data"][0]["id"]
    artist_name = data_artist["data"][0]["name"]

    # 2. Obtener top 50 canciones
    url_tracks = f"https://api.deezer.com/artist/{artist_id}/top"
    params_tracks = {"limit": 50}

    response_tracks = requests.get(url_tracks, params=params_tracks)
    data_tracks = response_tracks.json()

    df_tracks = pd.DataFrame(data_tracks["data"])

    if df_tracks.empty:
        print(f"No se encontraron canciones para: {artist_name}")
        return pd.DataFrame()

    # 3. Limpiar canciones
    df_tracks_limpio = pd.DataFrame()

    df_tracks_limpio["track_id"] = df_tracks["id"]
    df_tracks_limpio["track_title"] = df_tracks["title"]
    df_tracks_limpio["artist_id"] = artist_id
    df_tracks_limpio["artist_name"] = artist_name
    df_tracks_limpio["album_id"] = df_tracks["album"].apply(obtener_id)
    df_tracks_limpio["album_title"] = df_tracks["album"].apply(obtener_titulo)
    df_tracks_limpio["type"] = df_tracks["type"]

    # 4. Consultar solo álbumes únicos
    album_ids_unicos = df_tracks_limpio["album_id"].unique()

    albumes_info = []

    for album_id in album_ids_unicos:
        info_album = obtener_info_album(album_id)
        albumes_info.append(info_album)
        time.sleep(0.2)

    df_albums = pd.DataFrame(albumes_info)

    # 5. Unir canciones con info de álbum
    df_tracks_final = df_tracks_limpio.merge(
    df_albums,
    on="album_id",
    how="left")

    # convertir año a entero
    df_tracks_final["release_year"] = (
    df_tracks_final["release_year"].astype(int))

    return df_tracks_final
```
