```python
import pandas as pd
Rocio = pd.read_csv("artistas_completo_deezer_rocio.csv")
Teresa = pd.read_csv("artistas_completo_deezer_Teredin.csv")
Nati = pd.read_csv("artistas_completo_deezer_Nati.csv")
Cris = pd.read_csv("artistas_completo_Deezer_Cris.csv")
artistas_completo_deezer = pd.concat([Rocio, Teresa, Nati, Cris], ignore_index=True)
artistas_completo_deezer.to_csv("artistas_completo_deezer.csv", index=False, encoding="utf-8")
```
